```python
import pandas as pd
import os
```

```python
base_path = "data/splits"

paths = {
    "strategy_a_train": f"{base_path}/strategy_a/train.csv",
    "strategy_a_validation": f"{base_path}/strategy_a/validation.csv",
    "strategy_a_test": f"{base_path}/strategy_a/test.csv",
    "strategy_b_train": f"{base_path}/strategy_b/train.csv",
    "strategy_b_validation": f"{base_path}/strategy_b/validation.csv",
    "strategy_b_test": f"{base_path}/strategy_b/test.csv",
}

for name, path in paths.items():
    print(name, "->", os.path.exists(path))
```

```python
for name, path in paths.items():
    print("\nChecking:", name, path)
    try:
        temp_df = pd.read_csv(path)
        print("OK:", temp_df.shape)
    except Exception as e:
        print("ERROR:", e)
```

```python
dfs = {name: pd.read_csv(path) for name, path in paths.items()}

for name, df_split in dfs.items():
    print("\n" + "="*50)
    print(name)
    print("Shape:", df_split.shape)
    print("Columns:", df_split.columns.tolist())
    print("Label distribution:")
    print(df_split["label"].value_counts().sort_index())
```

```python
dfs["strategy_b_train"].head(3)
```

```python
def create_input_text(row):
    return (
        "Soru: " + str(row["soru"]).strip() + "\n"
        "Bağlam: " + str(row["context"]).strip() + "\n"
        "Cevap: " + str(row["cevap"]).strip()
    )

for name in dfs:
    dfs[name]["input_text"] = dfs[name].apply(create_input_text, axis=1)

for name, df_split in dfs.items():
    print(name, "-> input_text eklendi:", "input_text" in df_split.columns)
```

```python
print(dfs["strategy_b_train"].iloc[0]["input_text"])
print("\nLabel:", dfs["strategy_b_train"].iloc[0]["label"])
```

```python
for name in dfs:
    dfs[name]["input_char_length"] = dfs[name]["input_text"].str.len()
    dfs[name]["context_char_length"] = dfs[name]["context"].astype(str).str.len()
    dfs[name]["answer_char_length"] = dfs[name]["cevap"].astype(str).str.len()
    dfs[name]["question_char_length"] = dfs[name]["soru"].astype(str).str.len()

for name, df_split in dfs.items():
    print("\n" + "="*50)
    print(name)
    print(df_split[[
        "question_char_length",
        "context_char_length",
        "answer_char_length",
        "input_char_length"
    ]].describe(percentiles=[0.5, 0.75, 0.9, 0.95, 0.99]))
```

```python
MAX_CONTEXT_CHARS = 2500

def create_truncated_input_text(row, max_context_chars=MAX_CONTEXT_CHARS):
    question = str(row["soru"]).strip()
    context = str(row["context"]).strip()
    answer = str(row["cevap"]).strip()

    truncated_context = context[:max_context_chars]

    return (
        "Soru: " + question + "\n"
        "Bağlam: " + truncated_context + "\n"
        "Cevap: " + answer
    )

for name in dfs:
    dfs[name]["input_text_truncated"] = dfs[name].apply(create_truncated_input_text, axis=1)
    dfs[name]["input_truncated_char_length"] = dfs[name]["input_text_truncated"].str.len()

for name, df_split in dfs.items():
    print("\n" + "="*50)
    print(name)
    print(df_split["input_truncated_char_length"].describe(percentiles=[0.5, 0.75, 0.9, 0.95, 0.99]))
```

```python
longest_examples = dfs["strategy_b_train"].sort_values(
    "input_truncated_char_length",
    ascending=False
).head(5)

longest_examples[[
    "soru",
    "Score",
    "label",
    "question_char_length",
    "context_char_length",
    "answer_char_length",
    "input_truncated_char_length"
]]
```

```python
import os

os.makedirs("data/processed/strategy_a", exist_ok=True)
os.makedirs("data/processed/strategy_b", exist_ok=True)

def create_processed_df(df_split):
    return df_split[[
        "input_text_truncated",
        "label",
        "Score",
        "soru",
        "context",
        "cevap",
        "kaynak",
        "veri türü"
    ]].rename(columns={
        "input_text_truncated": "input_text"
    })

processed = {
    "strategy_a_train": create_processed_df(dfs["strategy_a_train"]),
    "strategy_a_validation": create_processed_df(dfs["strategy_a_validation"]),
    "strategy_a_test": create_processed_df(dfs["strategy_a_test"]),
    "strategy_b_train": create_processed_df(dfs["strategy_b_train"]),
    "strategy_b_validation": create_processed_df(dfs["strategy_b_validation"]),
    "strategy_b_test": create_processed_df(dfs["strategy_b_test"]),
}

processed["strategy_a_train"].to_csv("data/processed/strategy_a/train.csv", index=False)
processed["strategy_a_validation"].to_csv("data/processed/strategy_a/validation.csv", index=False)
processed["strategy_a_test"].to_csv("data/processed/strategy_a/test.csv", index=False)

processed["strategy_b_train"].to_csv("data/processed/strategy_b/train.csv", index=False)
processed["strategy_b_validation"].to_csv("data/processed/strategy_b/validation.csv", index=False)
processed["strategy_b_test"].to_csv("data/processed/strategy_b/test.csv", index=False)

print("Processed dosyalar kaydedildi.")
```
